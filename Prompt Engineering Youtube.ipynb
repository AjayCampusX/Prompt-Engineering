### SELF CONSISTENCY PROMPTING.

```python
import ollama 
from collections import Counter  #library to count occurance of each term.
import re
```

```python
System_Instruction= """You are a meticulous problem-solving AI. When presented with a question or problem, you must solve it extremely carefully.
           Walk through your logic step-by-step to ensure complete accuracy before reaching a conclusion. 
           After you have finished your reasoning, you must place your final answer on the very last line of your response, 
           strictly enclosed in double asterisks. Example: **[Your Final Answer]**"""
```

```python
Query= "What is the exact mathematical result of 10 + 20 / 5 * 2 - 3?"
```

```python
answers=[]

for i in range(5):
    output = ollama.generate(
        model='llama3',
        prompt=Query,
        system=System_Instruction, 
        options={'temperature': 0.6} 
    )
    
    text = output['response'].strip()
    
    # Simple regex to find content inside ** **
    match = re.search(r'\*\*(.*)\*\*', text)
    
    if match:
        result = match.group(1) #  using group(1)  one get the text inside the bold marks
        answers.append(result)
        print(f"Attempt {i+1}: {result}")
```

```python
final_answer = Counter(answers).most_common(1)[0][0]

print(f"\n🏆 Final Consensus: {final_answer}")
```
